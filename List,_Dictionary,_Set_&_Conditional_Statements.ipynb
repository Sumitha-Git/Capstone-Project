```python
#Create a list named age_list with five integer elements. For eg., [24, 25, 27, 28 ,29]
age_list = [24, 25, 27, 28, 29]
print(age_list)
```

```python
#Create a list named name_list with five string elements.
name_list = ["Kumar", "Sharmi", "Lakshmi", "Meera", "Prasanna"]
print(name_list)
```

```python
#Append the string "Yazhini" to name_list.
name_list = ["Kumar", "Sharmi", "Lakshmi", "Meera", "Prasanna"]
name_list.append('Raju')
print(name_list)
```

```python
#Insert the element 26 at index 2 in age_list.
age_list = [24, 25, 27, 28, 29]
age_list.insert(2, 26)
print(age_list)
```

```python
# Remove the string "Yazhini" from name_list.
name_list = ["Kumar", "Sharmi", "Lakshmi", "Meera", "Prasanna", "Yazhini"]
name_list.remove("Yazhini")
print(name_list)
```

```python
#Pop the last element from age_list.
age_list = [24, 25, 26, 27, 28, 29]
age_list.pop()
print(age_list)
```

```python
#Extend the age_list with additional ages [31,30,32].
age_list = [24, 25, 26, 27, 28]
age_list.extend([31, 30, 32])
print(age_list)
```

```python
#
Sort age_list in descending order.
age_list = [24, 25, 26, 27, 28, 31, 30, 32]
age_list.sort(reverse=True)
print(age_list)
```

```python
#Find the max age, min age, and sum of all ages from age_list.
age_list = [24, 25, 26, 27, 28, 31, 30, 32]

max_age = max(age_list)

min_age = min(age_list)

total_age = sum(age_list)

print("Maximum age:", max_age)
print("Minimum age:", min_age)
print("Sum of all ages:", total_age)
```

```python
#Create a set called my_set with the following values:  ['a','e','i','o','u','a','a','i']
#Analyze the output and explain the same.

my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

#The order may be different because sets are unordered.
print(my_set)
```

```python
#Attempt to access my_set[4]. If the code throws an error, provide an explanation.

my_set = {'a', 'e', 'i', 'o', 'u'}
print(my_set[4])

#When I try to access my_set[4], Python gives a TypeError: 'set' object is not subscriptable.
#This happens because sets are unordered collections and do not support indexing.
#Therefore, we cannot access set elements using index numbers like [4].
```

```python
#Create two sets:
#set1 with values: {1, 3, 5, 7, 9}
#set2 with values: {2, 3, 5, 8, 10}

set1 = {1, 3, 5, 7, 9}


set2 = {2, 3, 5, 8, 10}

# Print both sets
print("Set 1:", set1)
print("Set 2:", set2)
```

```python
#Compute and print the union and intersection of set1 and set2.

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

union_set = set1.union(set2)
intersection_set = set1.intersection(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```

```python
#Below Average: Score less than 4
score = int(input("Enter your score (0 to 10): "))
if score > 7:
    print("Performance Category: Above Average")
elif score >= 4 and score <= 7:
    print("Performance Category: Average")
elif score < 4:
    print("Performance Category: Below Average")
```
